```python
import can
import struct
import time
import threading

from pynput import keyboard


# ============================================================
# CubeMars AK45-10 V3.0 設定
# ============================================================

YAW_ID  = 105
ROLL_ID = 104

# 機械的な原点 [deg]
YAW_ORIGIN_DEG  = 180.0
ROLL_ORIGIN_DEG = 165.0

# 原点基準の可動範囲 [deg]
YAW_MIN_DEG  = -60.0
YAW_MAX_DEG  =  60.0

ROLL_MIN_DEG = -50.0
ROLL_MAX_DEG =  30.0


# ============================================================
# CAN設定
# ============================================================

CAN_PORT = "COM6@115200"
CAN_BITRATE = 1000000

bus = can.Bus(
    interface="slcan",
    channel=CAN_PORT,
    bitrate=CAN_BITRATE
)


# ============================================================
# Mode 1 : Current Loop
# ============================================================

CAN_PACKET_SET_CURRENT = 1


# ============================================================
# コンプライアンス制御設定
# ============================================================

# 仮想ばね
# 大きくすると硬くなる
YAW_KP  = 0.20      # [A/deg]
ROLL_KP = 0.40

# 仮想ダンパ
YAW_KD  = 0.003      # [A/(deg/s)]
ROLL_KD = 0.003

# 最大電流
# 最初はかなり低め推奨
YAW_MAX_CURRENT_A  = 1.00
ROLL_MAX_CURRENT_A = 1.30


# ============================================================
# 電流方向
#
# 目標方向と逆に動く場合は
# +1.0 → -1.0
# に変更
# ============================================================

YAW_CURRENT_SIGN  = 1.0
ROLL_CURRENT_SIGN = 1.0


# ============================================================
# 手動教示判定
# ============================================================

# 目標位置からこれ以上外れたら
# 手で動かされた可能性を見る
TEACH_ERROR_DEG = 3.0

# この角速度以上なら手動操作と判定
TEACH_VELOCITY_DEG_S = 3.0

# 手を離した判定
TEACH_RELEASE_VELOCITY_DEG_S = 0.5

# この時間動かなければTEACH終了
TEACH_RELEASE_TIME = 0.1

# 手動操作中の抵抗感
# 0なら完全にフリー
TEACH_DAMPING = 0.01

# 手動操作中の最大電流
TEACH_MAX_CURRENT_A = 1.0

# HOLD → TEACH 切替時のブレンド時間 [s]
TEACH_BLEND_TIME = 0.5

# TEACH → HOLD 復帰時のブレンド時間 [s]
HOLD_BLEND_TIME = 0.15


# ============================================================
# 制御周期
# ============================================================

CONTROL_INTERVAL = 0.005     # 200 Hz
DISPLAY_INTERVAL = 0.10     # 10 Hz


# ============================================================
# キーボード操作
# ============================================================

# 1回押すごとの最終目標角
KEY_STEP_DEG = 5.0

# キーボード目標値を実際の制御目標へ追従させる速度 [deg/s]
# 200Hzなら 15deg/s × 0.005s = 0.075deg/周期
TARGET_SLEW_RATE_DEG_S = 15.0
# ============================================================
# キーボード入力優先時間
# ============================================================

KEYBOARD_PRIORITY_TIME = 3.0  # [s]

last_keyboard_yaw_time = 0.0
last_keyboard_roll_time = 0.0

# ============================================================
# シリアル通信
#
# 必要ならTrue
# ============================================================

USE_SERIAL = True
SERIAL_PORT = "COM5"
SERIAL_BAUD = 115200

ser = None

if USE_SERIAL:

    import serial

    try:

        ser = serial.Serial(
            SERIAL_PORT,
            SERIAL_BAUD,
            timeout=0.1
        )

        print(
            f"Serial OPEN : {SERIAL_PORT}"
        )

    except Exception as e:

        print(
            "Serial OPEN ERROR:",
            e
        )

        ser = None


# ============================================================
# シリアル送信
# ============================================================

def send_serial(command):

    if ser is None:
        return

    if not ser.is_open:
        return

    try:

        data = f"{command}\r\n"

        ser.write(
            data.encode("utf-8")
        )

        ser.flush()

        print(
            f"\n[SERIAL TX] {command}"
        )

    except Exception as e:

        print(
            "\nSerial ERROR:",
            e
        )


# ============================================================
# CAN受信データ
# ============================================================

latest_angles = {
    YAW_ID: None,
    ROLL_ID: None
}

latest_angle_time = {
    YAW_ID: 0.0,
    ROLL_ID: 0.0
}

angle_lock = threading.Lock()

can_rx_running = False
can_rx_thread = None


# ============================================================
# CAN受信
# ============================================================

def can_receive_loop():

    global can_rx_running

    while can_rx_running:

        try:

            msg = bus.recv(
                timeout=0.1
            )

            if msg is None:
                continue

            function_id = (
                msg.arbitration_id >> 8
            ) & 0xFF

            motor_id = (
                msg.arbitration_id
                & 0xFF
            )

            # CubeMars状態フィードバック
            if function_id != 0x29:
                continue

            if motor_id not in latest_angles:
                continue

            if len(msg.data) != 8:
                continue

            # Position
            # int16
            # 0.1 deg / bit
            position_raw = int.from_bytes(
                msg.data[0:2],
                byteorder="big",
                signed=True
            )

            angle_deg = (
                position_raw * 0.1
            )

            with angle_lock:

                latest_angles[motor_id] = (
                    angle_deg
                )

                latest_angle_time[motor_id] = (
                    time.time()
                )

        except Exception as e:

            if can_rx_running:

                print(
                    "\nCAN RX ERROR:",
                    e
                )

                time.sleep(0.1)


# ============================================================
# CAN受信開始
# ============================================================

def start_can_receiver():

    global can_rx_running
    global can_rx_thread

    if can_rx_running:
        return

    can_rx_running = True

    can_rx_thread = threading.Thread(
        target=can_receive_loop,
        daemon=True,
        name="CubeMarsCANReceiver"
    )

    can_rx_thread.start()

    print(
        "CAN receiver START"
    )


# ============================================================
# CAN受信停止
# ============================================================

def stop_can_receiver():

    global can_rx_running

    can_rx_running = False

    print(
        "CAN receiver STOP"
    )


# ============================================================
# 最新角度取得
# ============================================================

def get_motor_angle(
    motor_id,
    max_age=0.5
):

    now = time.time()

    with angle_lock:

        angle = latest_angles.get(
            motor_id
        )

        timestamp = latest_angle_time.get(
            motor_id,
            0.0
        )

    if angle is None:
        return None

    # 古いデータは使用しない
    if (
        now - timestamp
        > max_age
    ):
        return None

    return angle


# ============================================================
# raw角度 → 原点基準角度
# ============================================================

def raw_to_relative(
    raw_angle,
    origin
):

    angle = (
        raw_angle
        - origin
        + 180.0
    ) % 360.0 - 180.0

    return angle


# ============================================================
# Mode 1 電流指令
# ============================================================

def send_current(
    motor_id,
    current_A
):

    current_A = float(
        current_A
    )

    # CubeMars仕様
    # A × 1000
    current_cmd = int(
        round(
            current_A
            * 1000.0
        )
    )

    # 念のためドライバ仕様上限
    current_cmd = max(
        -60000,
        min(
            60000,
            current_cmd
        )
    )

    can_id = (
        CAN_PACKET_SET_CURRENT << 8
    ) | motor_id

    data = struct.pack(
        ">i",
        current_cmd
    )

    msg = can.Message(
        arbitration_id=can_id,
        data=data,
        is_extended_id=True
    )

    bus.send(msg)


# ============================================================
# 目標角
# ============================================================

# キーボード等から指定される最終目標角
command_yaw_deg = 0.0
command_roll_deg = 0.0

# 実際にコンプライアンス制御へ使用する目標角
target_yaw_deg = 0.0
target_roll_deg = 0.0

target_lock = threading.Lock()


# ============================================================
# 手動教示状態
# ============================================================

control_state = {

    YAW_ID: {
        "prev_angle": None,
        "prev_time": None,
        "teaching": False,
        "last_motion_time": 0.0,
        "filtered_velocity": 0.0,
        "mode_blend": 1.0,
        "prev_teaching": False
    },

    ROLL_ID: {
        "prev_angle": None,
        "prev_time": None,
        "teaching": False,
        "last_motion_time": 0.0,
        "filtered_velocity": 0.0
    }

}


# ============================================================
# 目標角設定
# ============================================================

def set_head(
    yaw_deg,
    roll_deg
):
    global command_yaw_deg
    global command_roll_deg

    yaw_deg = float(yaw_deg)
    roll_deg = float(roll_deg)

    yaw_deg = max(YAW_MIN_DEG, min(YAW_MAX_DEG, yaw_deg))
    roll_deg = max(ROLL_MIN_DEG, min(ROLL_MAX_DEG, roll_deg))

    with target_lock:
        command_yaw_deg = yaw_deg
        command_roll_deg = roll_deg

    print(
        f"COMMAND "
        f"YAW={yaw_deg:+.1f}° "
        f"ROLL={roll_deg:+.1f}°"
    )


def move_toward(current, target, max_step):
    if target > current:
        return min(current + max_step, target)
    if target < current:
        return max(current - max_step, target)
    return current


# ============================================================
# 現在位置を目標にする
# ============================================================

def set_current_position_as_target():

    global command_yaw_deg
    global command_roll_deg
    global target_yaw_deg
    global target_roll_deg

    yaw_raw = get_motor_angle(YAW_ID)
    roll_raw = get_motor_angle(ROLL_ID)

    if yaw_raw is None or roll_raw is None:
        return False

    yaw = raw_to_relative(yaw_raw, YAW_ORIGIN_DEG)
    roll = raw_to_relative(roll_raw, ROLL_ORIGIN_DEG)

    with target_lock:
        command_yaw_deg = yaw
        command_roll_deg = roll
        target_yaw_deg = yaw
        target_roll_deg = roll

    print(
        f"初期目標 "
        f"YAW={yaw:+.1f}° "
        f"ROLL={roll:+.1f}°"
    )

    return True


# ============================================================
# 1軸コンプライアンス制御
# ============================================================

def compliance_control_axis(
    motor_id,
    current_angle,
    target_angle,
    kp,
    kd,
    max_current,
    current_sign,
    keyboard_priority=False
):

    state = control_state[motor_id]
    now = time.time()

    if state["prev_angle"] is None:
        state["prev_angle"] = current_angle
        state["prev_time"] = now
        state["filtered_velocity"] = 0.0
        state["mode_blend"] = 1.0
        state["prev_teaching"] = False

        return (0.0, False, 0.0)

    dt = now - state["prev_time"]
    if dt <= 0:
        dt = 0.001

    delta = current_angle - state["prev_angle"]
    raw_velocity = delta / dt

    alpha = 0.25

    state["filtered_velocity"] = (
        alpha * raw_velocity
        + (1.0 - alpha) * state["filtered_velocity"]
    )

    velocity = state["filtered_velocity"]
    error = target_angle - current_angle

    # HOLD用電流
    hold_current = kp * error - kd * velocity
    hold_current = max(
        -max_current,
        min(max_current, hold_current)
    )

    # TEACH用電流：バネ成分なし、粘性抵抗のみ
    teach_current = -TEACH_DAMPING * velocity
    teach_current = max(
        -TEACH_MAX_CURRENT_A,
        min(TEACH_MAX_CURRENT_A, teach_current)
    )
    # ============================================================
    # キーボード優先
    # ============================================================

    if keyboard_priority:

        # TEACH中でも強制的にHOLDへ戻す
        state["teaching"] = False

        # HOLDへ滑らかに移行
        state["mode_blend"] = max(
            0.0,
            state["mode_blend"]
        )
    # TEACH開始判定
    if not state["teaching"]:

        # キーボード操作後しばらくはTEACHへ移行しない
        if not keyboard_priority:

            if (
                abs(error) >= TEACH_ERROR_DEG
                and
                abs(velocity) >= TEACH_VELOCITY_DEG_S
            ):

                state["teaching"] = True

                state["last_motion_time"] = now

                state["mode_blend"] = 0.0

    if state["teaching"]:

        if abs(velocity) >= TEACH_RELEASE_VELOCITY_DEG_S:
            state["last_motion_time"] = now

        # HOLD → TEACH を滑らかにブレンド
        if TEACH_BLEND_TIME <= 0:
            state["mode_blend"] = 1.0
        else:
            state["mode_blend"] = min(
                1.0,
                state["mode_blend"] + dt / TEACH_BLEND_TIME
            )

        blend = state["mode_blend"]

        current_cmd = (
            (1.0 - blend) * hold_current
            + blend * teach_current
        )

        if (
            now - state["last_motion_time"]
            >= TEACH_RELEASE_TIME
        ):
            state["teaching"] = False
            state["mode_blend"] = 1.0

    else:

        # TEACH → HOLD も滑らかに戻す
        if state["mode_blend"] > 0.0:
            if HOLD_BLEND_TIME <= 0:
                state["mode_blend"] = 0.0
            else:
                state["mode_blend"] = max(
                    0.0,
                    state["mode_blend"] - dt / HOLD_BLEND_TIME
                )

            blend = state["mode_blend"]

            current_cmd = (
                (1.0 - blend) * hold_current
                + blend * teach_current
            )
        else:
            current_cmd = hold_current

    current_cmd *= current_sign

    state["prev_angle"] = current_angle
    state["prev_time"] = now
    state["prev_teaching"] = state["teaching"]

    return (
        current_cmd,
        state["teaching"],
        velocity
    )


# ============================================================
# システム状態
# ============================================================

system_running = False
control_thread = None


# ============================================================
# 2軸制御ループ
# ============================================================

def head_control_loop():

    global system_running
    global command_yaw_deg
    global command_roll_deg
    global target_yaw_deg
    global target_roll_deg

    last_display = 0.0

    print(
        "Compliance control START"
    )


    while system_running:

        loop_start = time.time()

        now = time.time()

        yaw_keyboard_priority = (
            now - last_keyboard_yaw_time
            < KEYBOARD_PRIORITY_TIME
        )

        roll_keyboard_priority = (
            now - last_keyboard_roll_time
            < KEYBOARD_PRIORITY_TIME
        )

        # ----------------------------------------------------
        # 現在角度
        # ----------------------------------------------------

        yaw_raw = get_motor_angle(
            YAW_ID
        )

        roll_raw = get_motor_angle(
            ROLL_ID
        )


        # フィードバックが無ければ停止
        if (
            yaw_raw is None
            or roll_raw is None
        ):

            send_current(
                YAW_ID,
                0.0
            )

            send_current(
                ROLL_ID,
                0.0
            )

            time.sleep(
                CONTROL_INTERVAL
            )

            continue


        # ----------------------------------------------------
        # 機械原点基準へ変換
        # ----------------------------------------------------

        yaw = raw_to_relative(
            yaw_raw,
            YAW_ORIGIN_DEG
        )

        roll = raw_to_relative(
            roll_raw,
            ROLL_ORIGIN_DEG
        )


        # ----------------------------------------------------
        # 目標角ランプ
        # ----------------------------------------------------

        with target_lock:

            max_step = (
                TARGET_SLEW_RATE_DEG_S
                * CONTROL_INTERVAL
            )

            target_yaw_deg = move_toward(
                target_yaw_deg,
                command_yaw_deg,
                max_step
            )

            target_roll_deg = move_toward(
                target_roll_deg,
                command_roll_deg,
                max_step
            )

            yaw_target = target_yaw_deg
            roll_target = target_roll_deg


        # ----------------------------------------------------
        # YAW制御
        # ----------------------------------------------------

        (
            yaw_current_cmd,
            yaw_teaching,
            yaw_velocity

        ) = compliance_control_axis(

            YAW_ID,

            yaw,

            yaw_target,

            YAW_KP,
            YAW_KD,

            YAW_MAX_CURRENT_A,

            YAW_CURRENT_SIGN,

            keyboard_priority=yaw_keyboard_priority
        )


        # ----------------------------------------------------
        # ROLL制御
        # ----------------------------------------------------

        (
            roll_current_cmd,
            roll_teaching,
            roll_velocity

        ) = compliance_control_axis(

            ROLL_ID,

            roll,

            roll_target,

            ROLL_KP,
            ROLL_KD,

            ROLL_MAX_CURRENT_A,

            ROLL_CURRENT_SIGN,

            keyboard_priority=roll_keyboard_priority
        )


        # ----------------------------------------------------
        # 手動教示中は
        # 現在位置を目標位置へ追従
        # ----------------------------------------------------

        with target_lock:

            if yaw_teaching:

                taught_yaw = max(
                    YAW_MIN_DEG,
                    min(YAW_MAX_DEG, yaw)
                )

                target_yaw_deg = taught_yaw
                command_yaw_deg = taught_yaw


            if roll_teaching:

                taught_roll = max(
                    ROLL_MIN_DEG,
                    min(ROLL_MAX_DEG, roll)
                )

                target_roll_deg = taught_roll
                command_roll_deg = taught_roll


            yaw_target_display = (
                target_yaw_deg
            )

            roll_target_display = (
                target_roll_deg
            )


        # ----------------------------------------------------
        # ソフトリミット
        # ----------------------------------------------------

        if (
            yaw <= YAW_MIN_DEG
            and yaw_current_cmd < 0
        ):
            yaw_current_cmd = 0.0

        if (
            yaw >= YAW_MAX_DEG
            and yaw_current_cmd > 0
        ):
            yaw_current_cmd = 0.0


        if (
            roll <= ROLL_MIN_DEG
            and roll_current_cmd < 0
        ):
            roll_current_cmd = 0.0

        if (
            roll >= ROLL_MAX_DEG
            and roll_current_cmd > 0
        ):
            roll_current_cmd = 0.0


        # ----------------------------------------------------
        # Mode 1送信
        # ----------------------------------------------------

        send_current(
            YAW_ID,
            yaw_current_cmd
        )

        send_current(
            ROLL_ID,
            roll_current_cmd
        )


        # ----------------------------------------------------
        # 表示
        # ----------------------------------------------------

        if (
            time.time()
            - last_display
            >= DISPLAY_INTERVAL
        ):

            yaw_mode = (
                "TEACH"
                if yaw_teaching
                else "HOLD "
            )

            roll_mode = (
                "TEACH"
                if roll_teaching
                else "HOLD "
            )

            print(
                "\r"
                f"YAW[{yaw_mode}] "
                f"T={yaw_target_display:+6.1f}° "
                f"C={yaw:+6.1f}° "
                f"I={yaw_current_cmd:+5.2f}A"
                "   |   "
                f"ROLL[{roll_mode}] "
                f"T={roll_target_display:+6.1f}° "
                f"C={roll:+6.1f}° "
                f"I={roll_current_cmd:+5.2f}A",
                end=""
            )

            last_display = (
                time.time()
            )


        # ----------------------------------------------------
        # 周期調整
        # ----------------------------------------------------

        elapsed = (
            time.time()
            - loop_start
        )

        sleep_time = (
            CONTROL_INTERVAL
            - elapsed
        )

        if sleep_time > 0:

            time.sleep(
                sleep_time
            )


    # --------------------------------------------------------
    # 終了時トルクOFF
    # --------------------------------------------------------

    send_current(
        YAW_ID,
        0.0
    )

    send_current(
        ROLL_ID,
        0.0
    )

    print(
        "\nCompliance control STOP"
    )


# ============================================================
# キーボード
# ============================================================

keyboard_listener = None


def on_key_press(key):

    global command_yaw_deg
    global command_roll_deg
    global system_running

    global last_keyboard_yaw_time
    global last_keyboard_roll_time

    try:
        k = key.char.lower()

    except AttributeError:
        return

    with target_lock:
        yaw = command_yaw_deg
        roll = command_roll_deg

    now = time.time()

    if k == "a":

        yaw -= KEY_STEP_DEG

        # YAWをしばらくキーボード優先
        last_keyboard_yaw_time = now

    elif k == "d":

        yaw += KEY_STEP_DEG

        last_keyboard_yaw_time = now

    elif k == "w":

        roll -= KEY_STEP_DEG

        # ROLLをしばらくキーボード優先
        last_keyboard_roll_time = now

    elif k == "s":

        roll += KEY_STEP_DEG

        last_keyboard_roll_time = now

        # ----------------------------------------
        # シリアル
        # ----------------------------------------

    elif k == "1":
            
        send_serial("1")
        return

    elif k == "2":

        send_serial("0,1")
        return

    elif k == "3":

        send_serial("0,2")
        return
    elif k == "4":

        send_serial("0,3")
        return
    
    elif k == "5":

            send_serial("0,0")
            return

    elif k == "q":

        print("\nSTOP requested")

        system_running = False

        send_current(
            YAW_ID,
            0.0
        )

        send_current(
            ROLL_ID,
            0.0
        )

        return False

    else:
        return

    # 可動範囲
    yaw = max(
        YAW_MIN_DEG,
        min(
            YAW_MAX_DEG,
            yaw
        )
    )

    roll = max(
        ROLL_MIN_DEG,
        min(
            ROLL_MAX_DEG,
            roll
        )
    )

    set_head(
        yaw,
        roll
    )


# ============================================================
# キーボード開始
# ============================================================

def start_keyboard():

    global keyboard_listener

    if (
        keyboard_listener is not None
        and keyboard_listener.is_alive()
    ):
        return

    keyboard_listener = (
        keyboard.Listener(
            on_press=on_key_press
        )
    )

    keyboard_listener.start()

    print(
        "Keyboard START"
    )


# ============================================================
# システム開始
# ============================================================

def start_system():

    global system_running
    global control_thread

    if system_running:

        print(
            "すでに動作中です"
        )

        return


    # CAN受信開始
    start_can_receiver()


    # フィードバック待ち
    print(
        "角度フィードバック待機中..."
    )

    start_time = time.time()

    while True:

        yaw_raw = get_motor_angle(
            YAW_ID
        )

        roll_raw = get_motor_angle(
            ROLL_ID
        )

        if (
            yaw_raw is not None
            and roll_raw is not None
        ):
            break

        if (
            time.time()
            - start_time
            > 3.0
        ):

            print(
                "角度を取得できません"
            )

            return

        time.sleep(0.05)


    # --------------------------------------------------------
    # 起動時は現在位置を目標にする
    #
    # これにより起動しただけでは動かない
    # --------------------------------------------------------

    yaw = raw_to_relative(
        yaw_raw,
        YAW_ORIGIN_DEG
    )

    roll = raw_to_relative(
        roll_raw,
        ROLL_ORIGIN_DEG
    )

    global command_yaw_deg
    global command_roll_deg
    global target_yaw_deg
    global target_roll_deg

    with target_lock:
        command_yaw_deg = yaw
        command_roll_deg = roll
        target_yaw_deg = yaw
        target_roll_deg = roll

    print(
        f"初期目標 YAW={yaw:+.1f}° "
        f"ROLL={roll:+.1f}°"
    )


    # 制御開始
    system_running = True

    control_thread = threading.Thread(
        target=head_control_loop,
        daemon=True,
        name="CubeMarsComplianceControl"
    )

    control_thread.start()


    # キーボード開始
    start_keyboard()


    print("")
    print("===================================")
    print(" Compliance Head Control")
    print("===================================")
    print("A : YAW -")
    print("D : YAW +")
    print("W : ROLL -")
    print("S : ROLL +")
    print("1 : Serial 0,1")
    print("2 : Serial 0,2")
    print("3 : Serial 0,3")
    print("Q : STOP")
    print("")
    print("手で動かす → TEACH")
    print("手を離す   → その角度をHOLD")
    print("===================================")


# ============================================================
# システム停止
# ============================================================

def stop_system():

    global system_running

    system_running = False

    send_current(
        YAW_ID,
        0.0
    )

    send_current(
        ROLL_ID,
        0.0
    )

    print(
        "\nSYSTEM STOP"
    )
```

システム起動

```python
start_system()
```
