```python
import can
import struct
import math
import time
# ============================================================
# CubeMars AK45-10 V3.0 設定
# ============================================================

# モータID
YAW_ID  = 105
ROLL_ID = 104

# 機械的な原点 [deg]
YAW_ORIGIN_DEG  = 145.0
ROLL_ORIGIN_DEG = 165.0

# 可動範囲 [deg]：原点から見た角度
YAW_MIN_DEG  = -60.0
YAW_MAX_DEG  =  60.0

ROLL_MIN_DEG = -50.0
ROLL_MAX_DEG =  30.0

# 出力軸最大回転速度 [rpm]
MAX_OUTPUT_RPM = 50.0

# AK45-10 V3.0
GEAR_RATIO = 10.0
POLE_PAIRS = 14

# CubeMars CAN command
CAN_PACKET_SET_POS_SPD = 6


# ============================================================
# CAN初期化
# ============================================================
# ↓使用しているUSB-CANに合わせて変更してください
#
# Linux SocketCANなら:
#
# bus = can.Bus(
#     interface="socketcan",
#     channel="can0",
#     bitrate=1000000
# )
#
# PCANなら:
#
# bus = can.Bus(
#     interface="pcan",
#     channel="PCAN_USBBUS1",
#     bitrate=1000000
# )

import can

bus = can.Bus(
    interface="slcan",
    channel="COM6@115200",
    bitrate=1000000
)
```

```python
def get_motor_angle(motor_id, timeout=1.0):
    """
    CubeMars AK45-10 V3.0
    指定モータの現在角度を取得する

    Parameters
    ----------
    motor_id : int
        モータID
        例:
            YAW  = 105
            ROLL = 104

    timeout : float
        受信待ち時間 [s]

    Returns
    -------
    float
        現在角度 [deg]

    None
        timeoutまでに受信できなかった場合
    """

    start_time = time.time()

    while time.time() - start_time < timeout:

        msg = bus.recv(timeout=0.1)

        if msg is None:
            continue

        # Extended CAN ID
        function_id = (msg.arbitration_id >> 8) & 0xFF
        rx_motor_id = msg.arbitration_id & 0xFF

        # ----------------------------------------
        # CubeMars 状態フィードバック
        # Function ID = 0x29
        # ----------------------------------------
        if function_id != 0x29:
            continue

        # 対象モータのみ
        if rx_motor_id != motor_id:
            continue

        # 8 byteでなければ無視
        if len(msg.data) != 8:
            continue

        # ----------------------------------------
        # Position
        # byte 0-1
        # signed int16
        # 0.1 deg / bit
        # ----------------------------------------
        position_raw = int.from_bytes(
            msg.data[0:2],
            byteorder="big",
            signed=True
        )

        angle_deg = position_raw * 0.1

        return angle_deg

    return None

# ============================================================
# 出力軸 rpm → ERPM
# ============================================================

def output_rpm_to_erpm(output_rpm):
    """
    出力軸の回転速度[rpm]をCubeMarsで使用するERPMへ変換
    """

    return output_rpm * GEAR_RATIO * POLE_PAIRS


# ============================================================
# CubeMars Position-Speed Loop送信
# ============================================================

def send_position_speed(
    motor_id,
    motor_angle_deg,
    output_speed_rpm=10.0,
    output_accel_rpm_s=100.0
):
    """
    CubeMars AK V3
    Position-Speed Loop Mode

    motor_id:
        CAN ID

    motor_angle_deg:
        CubeMars絶対角度 [deg]

    output_speed_rpm:
        出力軸最大速度 [rpm]

    output_accel_rpm_s:
        出力軸加速度 [rpm/s]
    """

    # --------------------------------------------------------
    # 速度制限
    # --------------------------------------------------------

    output_speed_rpm = abs(float(output_speed_rpm))

    if output_speed_rpm > MAX_OUTPUT_RPM:
        print(
            f"速度 {output_speed_rpm:.1f} rpm は上限を超えています。"
            f"{MAX_OUTPUT_RPM:.1f} rpm に制限します。"
        )

        output_speed_rpm = MAX_OUTPUT_RPM


    # --------------------------------------------------------
    # 出力軸rpm → ERPM
    # --------------------------------------------------------

    speed_erpm = output_rpm_to_erpm(output_speed_rpm)

    accel_erpm_s = (
        output_accel_rpm_s
        * GEAR_RATIO
        * POLE_PAIRS
    )


    # --------------------------------------------------------
    # CubeMars CANデータ変換
    #
    # position : deg × 10000
    #
    # speed:
    #   送信値 = ERPM / 10
    #
    # acceleration:
    #   送信値 = ERPM/s / 10
    # --------------------------------------------------------

    position_cmd = int(round(motor_angle_deg * 10000.0))

    speed_cmd = int(round(speed_erpm))

    accel_cmd = int(round(accel_erpm_s))


    # int16範囲チェック
    speed_cmd = max(-32768, min(32767, speed_cmd))
    accel_cmd = max(-32768, min(32767, accel_cmd))


    # --------------------------------------------------------
    # CAN Extended ID
    #
    # mode = 6
    # motor ID = 下位8bit
    # --------------------------------------------------------

    can_id = (
        (CAN_PACKET_SET_POS_SPD << 8)
        | motor_id
    )


    # --------------------------------------------------------
    # 8 byte CAN DATA
    #
    # position : int32
    # speed    : int16
    # accel    : int16
    #
    # Big Endian
    # --------------------------------------------------------

    data = struct.pack(
        ">ihh",
        position_cmd,
        speed_cmd,
        accel_cmd
    )


    message = can.Message(
        arbitration_id=can_id,
        data=data,
        is_extended_id=True
    )


    bus.send(message)
    print(
    f"[TX] motor={motor_id} "
    f"angle={motor_angle_deg:.1f}deg "
    f"CAN_ID=0x{can_id:X} "
    f"DATA={data.hex(' ')} "
    f"speed_cmd={speed_cmd} "
    f"accel_cmd={accel_cmd}"
    )

def move_yaw(angle_deg,speed_rpm=10.0,accel_rpm_s=100.0):
    """
    ヨー軸を指定角度へ移動

    angle_deg:
        原点基準角度 [deg]
        -60 ～ +60

    speed_rpm:
        出力軸速度 [rpm]
        0 ～ 50

    例:
        move_yaw(30)
        → 原点145°から+30°
        → CubeMars絶対角度175°
    """

    angle_deg = float(angle_deg)

    # ----------------------------
    # 可動範囲チェック
    # ----------------------------

    if not (YAW_MIN_DEG<= angle_deg<= YAW_MAX_DEG):

        raise ValueError(
            f"ヨー角度が範囲外です: {angle_deg:.1f}° "
            f"(許容範囲 {YAW_MIN_DEG:.1f}～"
            f"{YAW_MAX_DEG:.1f}°)"
        )


    # ----------------------------
    # 実際にCubeMarsへ送る角度
    # ----------------------------

    motor_angle = (
        YAW_ORIGIN_DEG
        + angle_deg
    )


    send_position_speed(
        YAW_ID,
        motor_angle,
        speed_rpm,
        accel_rpm_s
    )


    print(
        f"YAW : {angle_deg:+.1f}° "
        f"(Motor={motor_angle:.1f}°) "
        f"Speed={min(abs(speed_rpm), MAX_OUTPUT_RPM):.1f} rpm"
    )

def move_roll(angle_deg, speed_rpm=10.0, accel_rpm_s=100.0):
    """
    ロール軸を指定角度へ移動

    angle_deg:
        原点基準角度 [deg]
        -50 ～ +30

    speed_rpm:
        出力軸速度 [rpm]
        0 ～ 50
    """

    angle_deg = float(angle_deg)


    # ----------------------------
    # 可動範囲チェック
    # ----------------------------

    if not (ROLL_MIN_DEG<= angle_deg <= ROLL_MAX_DEG):

        raise ValueError(
            f"ロール角度が範囲外です: {angle_deg:.1f}° "
            f"(許容範囲 {ROLL_MIN_DEG:.1f}～"
            f"{ROLL_MAX_DEG:.1f}°)"
        )


    # ----------------------------
    # 実際にCubeMarsへ送る角度
    # ----------------------------

    motor_angle = (
        ROLL_ORIGIN_DEG
        + angle_deg
    )


    send_position_speed(
        ROLL_ID,
        motor_angle,
        speed_rpm,
        accel_rpm_s
    )


    print(
        f"ROLL: {angle_deg:+.1f}° "
        f"(Motor={motor_angle:.1f}°) "
        f"Speed={min(abs(speed_rpm), MAX_OUTPUT_RPM):.1f} rpm"
    )

def move_head(yaw_deg, roll_deg, yaw_speed_rpm=10.0, roll_speed_rpm=10.0, accel_rpm_s=100.0):
    """
    ヨー・ロールを同時に指定

    yaw_deg:
        -60 ～ +60 deg

    roll_deg:
        -50 ～ +30 deg
    """

    # ----------------------------
    # 指令値チェック
    # ----------------------------

    if not YAW_MIN_DEG <= yaw_deg <= YAW_MAX_DEG:
        raise ValueError(
            f"YAW範囲外: {yaw_deg}°"
        )

    if not ROLL_MIN_DEG <= roll_deg <= ROLL_MAX_DEG:
        raise ValueError(
            f"ROLL範囲外: {roll_deg}°"
        )


    # ----------------------------
    # 2軸送信
    # ----------------------------

    move_yaw(
        yaw_deg,
        yaw_speed_rpm,
        accel_rpm_s
    )

    move_roll(
        roll_deg,
        roll_speed_rpm,
        accel_rpm_s
    )

def hold_position(
    motor_id,
    target_angle_deg,
    speed_rpm=1.0,
    accel_rpm_s=20.0,
    interval=0.05
    ):
    try:
        while True:
            send_position_speed(
                motor_id,
                target_angle_deg,
                speed_rpm,
                accel_rpm_s
            )

            current = get_motor_angle(
                motor_id,
                timeout=0.1
            )

            if current is not None:
                error = target_angle_deg - current
                print(
                    f"\rtarget={target_angle_deg:.1f}° "
                    f"current={current:.1f}° "
                    f"error={error:+.1f}°",
                    end=""
                )

            time.sleep(interval)

    except KeyboardInterrupt:
        print("\n保持終了")

def hold_head(
    yaw_deg,
    roll_deg,
    yaw_speed_rpm=0.5,
    roll_speed_rpm=0.5,
    accel_rpm_s=20.0,
    interval=0.05
):
    """
    YAWとROLLを同時に指定して、その位置を保持する

    yaw_deg:
        原点基準角度 [-60 ～ +60 deg]

    roll_deg:
        原点基準角度 [-50 ～ +30 deg]

    yaw_speed_rpm:
        YAW出力軸速度 [rpm]

    roll_speed_rpm:
        ROLL出力軸速度 [rpm]

    accel_rpm_s:
        出力軸加速度 [rpm/s]

    interval:
        位置指令再送間隔 [s]

    Ctrl+C で終了
    """

    # ==========================================
    # 可動範囲チェック
    # ==========================================

    if not YAW_MIN_DEG <= yaw_deg <= YAW_MAX_DEG:
        raise ValueError(
            f"YAW範囲外: {yaw_deg}° "
            f"({YAW_MIN_DEG}～{YAW_MAX_DEG}°)"
        )

    if not ROLL_MIN_DEG <= roll_deg <= ROLL_MAX_DEG:
        raise ValueError(
            f"ROLL範囲外: {roll_deg}° "
            f"({ROLL_MIN_DEG}～{ROLL_MAX_DEG}°)"
        )

    # ==========================================
    # モータ絶対角度へ変換
    # ==========================================

    yaw_target = YAW_ORIGIN_DEG + yaw_deg
    roll_target = ROLL_ORIGIN_DEG + roll_deg

    print(
        f"YAW  target = {yaw_deg:+.1f}° "
        f"(motor={yaw_target:.1f}°)"
    )

    print(
        f"ROLL target = {roll_deg:+.1f}° "
        f"(motor={roll_target:.1f}°)"
    )

    print("位置保持開始 / Ctrl+Cで終了")

    try:

        while True:

            # ==================================
            # YAW目標送信
            # ==================================

            send_position_speed(
                YAW_ID,
                yaw_target,
                yaw_speed_rpm,
                accel_rpm_s
            )

            # ==================================
            # ROLL目標送信
            # ==================================

            send_position_speed(
                ROLL_ID,
                roll_target,
                roll_speed_rpm,
                accel_rpm_s
            )

            # ==================================
            # 現在角取得
            # ==================================

            yaw_current = get_motor_angle(
                YAW_ID,
                timeout=0.1
            )

            roll_current = get_motor_angle(
                ROLL_ID,
                timeout=0.1
            )

            # ==================================
            # 表示
            # ==================================

            if yaw_current is not None:
                yaw_error = yaw_target - yaw_current
            else:
                yaw_error = None

            if roll_current is not None:
                roll_error = roll_target - roll_current
            else:
                roll_error = None


            if (
                yaw_current is not None
                and roll_current is not None
            ):

                print(
                    f"\r"
                    f"YAW "
                    f"T={yaw_target:6.1f}° "
                    f"C={yaw_current:6.1f}° "
                    f"E={yaw_error:+5.1f}°  |  "
                    f"ROLL "
                    f"T={roll_target:6.1f}° "
                    f"C={roll_current:6.1f}° "
                    f"E={roll_error:+5.1f}°",
                    end=""
                )

            time.sleep(interval)

    except KeyboardInterrupt:

        print("\n位置保持終了")
import threading
import time

# ============================================================
# 共有する目標値
# ============================================================

target_yaw_deg = 0.0
target_roll_deg = 0.0

target_lock = threading.Lock()

head_hold_running = False
head_hold_thread = None


# ============================================================
# 目標値更新
# ============================================================

def set_head(yaw_deg, roll_deg):
    global target_yaw_deg, target_roll_deg

    if not YAW_MIN_DEG <= yaw_deg <= YAW_MAX_DEG:
        raise ValueError(
            f"YAW範囲外: {yaw_deg}° "
            f"({YAW_MIN_DEG}～{YAW_MAX_DEG}°)"
        )

    if not ROLL_MIN_DEG <= roll_deg <= ROLL_MAX_DEG:
        raise ValueError(
            f"ROLL範囲外: {roll_deg}° "
            f"({ROLL_MIN_DEG}～{ROLL_MAX_DEG}°)"
        )

    with target_lock:
        target_yaw_deg = float(yaw_deg)
        target_roll_deg = float(roll_deg)


# ============================================================
# 保持スレッド
# ============================================================

def head_hold_loop(
    yaw_speed_rpm=0.5,
    roll_speed_rpm=0.5,
    accel_rpm_s=20.0,
    interval=0.05
):
    global head_hold_running

    print("HEAD HOLD START")

    while head_hold_running:

        with target_lock:
            yaw_deg = target_yaw_deg
            roll_deg = target_roll_deg

        yaw_target = YAW_ORIGIN_DEG + yaw_deg
        roll_target = ROLL_ORIGIN_DEG + roll_deg

        # YAW
        send_position_speed(
            YAW_ID,
            yaw_target,
            yaw_speed_rpm,
            accel_rpm_s
        )

        # ROLL
        send_position_speed(
            ROLL_ID,
            roll_target,
            roll_speed_rpm,
            accel_rpm_s
        )

        time.sleep(interval)

    print("HEAD HOLD STOP")


# ============================================================
# 保持開始
# ============================================================

def start_head_hold(
    yaw_speed_rpm=0.5,
    roll_speed_rpm=0.5,
    accel_rpm_s=20.0,
    interval=0.05
):
    global head_hold_running
    global head_hold_thread

    if head_hold_running:
        print("すでにHEAD HOLDは動作中です")
        return

    head_hold_running = True

    head_hold_thread = threading.Thread(
        target=head_hold_loop,
        kwargs={
            "yaw_speed_rpm": yaw_speed_rpm,
            "roll_speed_rpm": roll_speed_rpm,
            "accel_rpm_s": accel_rpm_s,
            "interval": interval
        },
        daemon=True
    )

    head_hold_thread.start()


# ============================================================
# 保持停止
# ============================================================

def stop_head_hold():
    global head_hold_running

    head_hold_running = False
```

```python
import threading
import serial
from pynput import keyboard

# ============================================================
# キーボード操作設定
# ============================================================

KEY_STEP_DEG = 2.0

keyboard_yaw = 0.0
keyboard_roll = 0.0

keyboard_running = False
keyboard_listener = None

keyboard_lock = threading.Lock()


# ============================================================
# シリアル通信設定
# ============================================================

SERIAL_PORT = "COM5"     # 実際のArduino等のCOM番号に変更
SERIAL_BAUD = 115200


ser = serial.Serial(
    SERIAL_PORT,
    SERIAL_BAUD,
    timeout=0.1
)

print("Serial connected:", SERIAL_PORT)


# ============================================================
# シリアル送信
# ============================================================

def send_serial(command):

    data = f"{command}\r\n"

    ser.write(
        data.encode("utf-8")
    )

    print(f"\n[SERIAL TX] {command}")


# ============================================================
# キーボード入力処理
# ============================================================

def on_key_press(key):

    global keyboard_yaw
    global keyboard_roll
    global keyboard_running

    try:

        k = key.char.lower()

    except AttributeError:

        return

    with keyboard_lock:

        # ----------------------------------------
        # YAW
        # ----------------------------------------

        if k == "a":

            keyboard_yaw -= KEY_STEP_DEG

        elif k == "d":

            keyboard_yaw += KEY_STEP_DEG


        # ----------------------------------------
        # ROLL
        # Wがマイナス方向
        # ----------------------------------------

        elif k == "w":

            keyboard_roll -= KEY_STEP_DEG

        elif k == "s":

            keyboard_roll += KEY_STEP_DEG


        # ----------------------------------------
        # シリアル
        # ----------------------------------------

        elif k == "1":

            send_serial("1")
            return

        elif k == "2":

            send_serial("0,1")
            return

        elif k == "3":

            send_serial("0,2")
            return
        elif k == "4":

            send_serial("0,3")
            return
        elif k == "5":

            send_serial("0,0")
            return

        # ----------------------------------------
        # 終了
        # ----------------------------------------

        elif k == "q":

            keyboard_running = False

            print("\nKeyboard control STOP")
            stop_head_hold()

            return False


        else:

            return


        # ========================================
        # 可動範囲制限
        # ========================================

        keyboard_yaw = max(
            YAW_MIN_DEG,
            min(
                YAW_MAX_DEG,
                keyboard_yaw
            )
        )

        keyboard_roll = max(
            ROLL_MIN_DEG,
            min(
                ROLL_MAX_DEG,
                keyboard_roll
            )
        )


        # ========================================
        # 新しい目標位置を保持スレッドへ渡す
        # ========================================

        set_head(
            keyboard_yaw,
            keyboard_roll
        )


        print(
            f"\r"
            f"YAW={keyboard_yaw:+6.1f}°  "
            f"ROLL={keyboard_roll:+6.1f}°     ",
            end=""
        )


# ============================================================
# キーボード制御開始
# ============================================================

def start_keyboard_control(
    initial_yaw=0.0,
    initial_roll=0.0
):

    global keyboard_yaw
    global keyboard_roll
    global keyboard_running
    global keyboard_listener


    keyboard_yaw = initial_yaw
    keyboard_roll = initial_roll


    # 最初の位置設定
    set_head(
        keyboard_yaw,
        keyboard_roll
    )


    keyboard_running = True


    keyboard_listener = keyboard.Listener(
        on_press=on_key_press
    )

    keyboard_listener.start()


    print(
        "Keyboard control START\n"
        "A/D : YAW -/+\n"
        "W/S : ROLL -/+\n"
        "1/2/3 : Serial TX\n"
        "Q : STOP"
    )


# ============================================================
# キーボード制御停止
# ============================================================

def stop_keyboard_control():

    global keyboard_running
    global keyboard_listener

    keyboard_running = False

    if keyboard_listener is not None:

        keyboard_listener.stop()

        keyboard_listener = None

    print("Keyboard control STOP")
```
